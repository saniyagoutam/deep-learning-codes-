```python
#SANIYA GOUTAM PRN-23070621126
import numpy as np
import matplotlib.pyplot as plt

```

```python
x=np.array([100,150,200,250,300])
y=np.array([20,25,30,35,40])
```

```python
xm=x.mean()
ym=y.mean()
print(xm,ym)
```

```python
b = np.sum((x - xm) * (y - ym))
c = np.sum((x - xm)**2)
b1 = b / c
b0 = ym - b1 * xm

print(f"Calculated Slope (b1): {b1}")
print(f"Calculated Y-intercept (b0): {b0}")
```

```python
y_pred = b0 + b1 * x
print(f"Original x values: {x}")
print(f"Original y values: {y}")
print(f"Predicted y values (y_pred): {y_pred}")
```

```python
difference = y - y_pred
print(difference)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print(f"(MAE): {mae}")
print(f"(MSE): {mse}")
print(f"(RMSE): {rmse}")
```

```python
plt.scatter(x, y, label='Original Data')
plt.plot(x, y_pred, color='red', label='Regression Line')
plt.xlabel('X (Independent Variable)')
plt.ylabel('Y (Dependent Variable)')
plt.grid(True)
plt.show()
```
